# 1. Create a 1D array of numbers from 50 to 100 with a step of 5.

```python
import numpy as np
arr=np.arange(50,101,5).reshape(11,1)
print(arr)
```

# 2. Given arr = np.array([[4,8],[15,16]]), print its .dtype and convert it to float32.

```python
arr=np.array([[4,8],[15,16]])
print(arr.dtype)
arr.dtype=float
print(arr.dtype)
```

# 3. Create a (3,3) array of random integers between 1-9 and print its transpose.

```python
arr=np.random.randint(1,9,9).reshape(3,3)
print(arr)
print(arr.T)
```

# 4. Given arr = np.arange(30).reshape(5,6), extract the last 2 columns using slicing.

```python
arr=np.arange(30).reshape(5,6)
print(arr)
print(arr[:,4:])
```

# 5. Use boolean indexing on np.array([12,45,7,89,23,56]) to get all values divisible by 3.

```python
arr=np.array([12,45,7,89,23,56])
print(arr[arr%3==0])
```

# 6. Given A = np.array([[2,1],[1,3]]) and B = np.array([[1,0],[0,1]]), compute A @ B.

```python
a=np.array([[2,1],[1,3]])
b=np.array([[1,0],[0,1]])
print(a @ b)
```

# 10. Create a (2,4) array and use .flatten() and .ravel() — modify each result and observe which one changes the original array.

```python
arr=np.arange(1,9).reshape(2,4)
print(arr)
x=arr.flatten() # it doesn't change the original array
print(x)

print(arr.ravel())  # it usually change the original array
```
