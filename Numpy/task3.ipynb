# Q1 create a program to make 1d array of 10 zeros

```python
import numpy as np
arr=np.zeros(10)
print(arr)
```

# Q2 write a program to create 3,6 matrix filled with digit 1

```python
arr=np.arange(1,19).reshape(3,6)
print(arr)
```

# Q3 convert a 2d array of  shape 2,6 into shape 3,4

```python
arr=np.arange(1,13).reshape(2,6)
print(arr)
print(arr.reshape(3,4))
```

# Q4 use arange function to genrate number from 0-20 with stepsize 2 and reshape in possible format

```python
arr=np.arange(0,21,2)
print(arr)
print(arr.reshape(1,11))
print(arr.reshape(11,1))
```

# Q5 create a 2d array of 4,4 matrix with random intiger numbers

```python
arr=np.random.randint(1,100,16).reshape(4,4)
print(arr)
```
