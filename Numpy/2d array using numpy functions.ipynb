# creating 2d array using numpy functions --- 16/02/2026

```python
import numpy as np
a=np.arange(1,11)
print(a)
print(type(a))
```

# reshape :- reshape converts into matrix format by using matrix format

```python
a=np.arange(1,10)
print(a.reshape(3,3))
```

```python
a=np.arange(1,10).reshape(3,3)
print(a)
```

```python
a=np.arange(1,82).reshape(9,9)
print(a)
```

# random :- it gives random values between 0-1

```python
n=np.random.random(25)
print(n)
```

```python
n=np.random.random(25).reshape(5,5)
print(n)
```

# randint:-

```python
a=np.random.randint(12,45,7)
print(a)
```

```python
a=np.random.randint(12,45,8).reshape(2,4 )
print(a)
```

```python
# create a array of random intiger numbers from 1-51 and reshape the array into possible format
a=np.random.randint(1,51,36)
print(a.reshape(6,6))
print(a.reshape(1,36))
print(a.reshape(36,1))
print(a.reshape(4,9))
print(a.reshape(9,4))
```

# zeros:-

```python
a=np.zeros(4)
print(a)
```

```python
a=np.zeros((2,3))
print(a)
```

```python
# create a array of zeros in matrix format 5,5 using reshape and without reshape
a=np.zeros(20).reshape(4,5)
b=np.zeros((4,5))
print(a)
print(b)
```

# ones:-

```python
a=np.ones(20).reshape(4,5)
b=np.ones((4,5))
print(a)
print(b)
```

# how to create identity matrix

```python
a=np.eye(4)
print(a)
```

# how to change data type of array

```python
arr=np.eye(4,dtype=int)
print(arr)
```

```python
arr=np.random.randint(12,34,(4,8))   # (4,8) is a tuple that create 4X8 matrix
print(arr)
```

```python
arr=np.random.randint(1,100,20).reshape(4,5)
print(arr)
```

```python
print(arr[3,4])
```

# Boolean indexing :- filters data based on condition

```python
print(arr[arr>50])
print(arr[arr%2==0])
```
