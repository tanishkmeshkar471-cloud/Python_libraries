```python
import time
import numpy as np

size = 10000

#list
a = list(range(size))

start_1 = time.time()
a = list(map(lambda i : i**2,a))
end_1 = time.time() - start_1

#arry
a = np.arange(size)
start_a = time.time()
a = [a**2]
end_a = time.time() - start_a

print("Time taken ty list",end_1)
print("Time taken ty arrya",end_a)
```
