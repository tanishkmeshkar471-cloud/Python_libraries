# 2.Create a 3x3 array of all zeros, then a 3x3 identity matrix.

```python
arr=np.zeros((3,3),dtype=int)
print(arr)
print()
arr=np.eye(3,dtype=int)
print(arr)
```

# 3.Create np.arange(20), reshape it into a 4x5 array, and print its .ndim, .shape, and .size.

```python
arr=np.arange(20).reshape(4,5)
print(arr)
print(arr.ndim)
print(arr.shape)
print(arr.size)
```

# 7.Create a (2,3) array of ones and check its .itemsize and total memory (itemsize * size).

```python
arr=np.ones((2,3))
print(arr)
print(arr.itemsize)
print(arr.itemsize*arr.size)
```
