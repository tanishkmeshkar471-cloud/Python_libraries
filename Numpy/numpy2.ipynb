```python
import numpy as np
```

```python
# how to create array ----------------------ans  -> .array()

a=[12,13,14]
print(a)
print(type(a))
```

```python
#array using list

arr=np.array(a)
print(type(arr))
print(arr)    # numpy.ndarray  means n number of dimentions array
```

```python
arr2=np.array(list)
print(arr2)
print(type(arr2))
```

```python
arr=np.array(list(range(1,11)))
print(arr)
print(type(arr))
```

```python
#array attribute
# 1.  .ndim to check dimention
#2.   .shape  to show the shape
#3.   .size   shows the total number of array elements
#4.   .dtype
#5.   .itemsize  it shows how much memory single element takes
```

```python
print(arr.ndim)  #it shows the dimention of array
```

```python
print(arr.shape) #it shows the shape of array
```

```python
print(arr.size) #it shows the total number of array elements
```

```python
arr1=np.array([12,13,14])
arr1=np.array([12.1,13.0,14.9])
arr3=np.array([12,"77777777777777",67])
arr3=np.array([12,77777777777777,67,True,False])
print(arr3.dtype)                  # U unicode and uske baad ki value ki length
print(arr3)
                                  #dtype shows the data type of data contained by array
```

```python
str_array=np.array(["12",13,45])
float_array=np.array([12.45,78,760])
int_array=np.array([12,78,760])

print(str_array.itemsize)
print(float_array.itemsize)
print(int_array.itemsize)
```
