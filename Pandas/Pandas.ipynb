```python
import numpy as np
import pandas as pd

# pandas is used for the data manipulation task
```

Series -- 
it is basically a single column or row the dimention of series will be 1d so this is how we identify series

dataFrame:- it is a collection/combination of multiple/more than 1 series

```python
# 27-10-2026

# how we create series


a=pd.Series([12,45,67])
print(a)
print(a.dtype)
print(type(a))
```

```python
# how to create series using numpy array

arr1=np.array([1,2,3,4,5,6])
print(arr1.ndim)
print(type(arr1))

s=pd.Series(arr1)
print(type(s))
print(s)
```

```python
# to convert dictionaries in Series
dic={
    "marks":[30,50,60,70,90,100]
}

s=pd.Series(dic)
print(dic)
print(s)
print(type(s))
print(s.dtype)
```

```python

arr=np.arange(1,11)
s=pd.Series(arr)

print(arr)
print(s)
print(s.dtype)
print(s.shape)
print(s.size)
print(s.ndim)
```

```python
# how to give name to the series
s=pd.Series(arr,name="Tanishk's Series")
print(s)
```

```python
# how to give custom indexing
arr=np.arange(1,6)
s=pd.Series(arr,index=["A","B","C","D","E"])
print(s)
```

```python
arr1=np.arange(10,100,10)
arr2=np.arange(1,10)
s=pd.Series(arr2,index=arr1)
print(s)
```
